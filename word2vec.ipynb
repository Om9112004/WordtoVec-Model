## Applying NLP steps ###

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
potter=[]
data=sent_tokenize(data)
```

```python
len(data)
```

```python
for i in data:
    potter.append(simple_preprocess(i))
```

```python
potter
```

```python
from nltk.corpus import stopwords
```

```python
stopwords=set(stopwords.words('english'))
```

```python
stopwords
```

```python
def stopword_removal(text):
    cleaned_text=[]
    for sentence in text:
        new_sentence=[]
        for word in sentence:
            if word.lower() not in stopwords:
                new_sentence.append(word)
        cleaned_text.append(new_sentence)
    return cleaned_text
```

```python
cleaned_potter=stopword_removal(potter)
```

```python
import gensim
```

```python
models=gensim.models.Word2Vec(window=10,min_count=2)
```

```python
models
```

```python
models.build_vocab(cleaned_potter)
```

```python
models.train(cleaned_potter,total_examples=models.corpus_count,epochs=models.epochs)
```
